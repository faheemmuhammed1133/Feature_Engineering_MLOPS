```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 

df = pd.read_csv("Datasets/customer_churn.csv")
print(df.shape)
df.head
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,roc_auc_score,classification_report

legitmat_features  = ["tenure_months","monthly_charges","contract_type","support_calls"]

X_clean = pd.get_dummies(df[legitmat_features],columns=["contract_type"],drop_first=True)
y=df["churn"]

X_train, X_test, y_train, y_test = train_test_split(
   X_clean,y,test_size=.25,random_state=42,stratify=y
)

model_clean = LogisticRegression(max_iter=1000).fit(X_train,y_train)

acc_clean = accuracy_score(y_test,model_clean.predict(X_test))
auc_clean = roc_auc_score(y_test,model_clean.predict_proba(X_test)[:,1])

print(acc_clean*100)
auc_clean*100
```

```python
df_leaked = df.copy()
df_leaked["days_since_cancellation"]= df_leaked["days_since_cancellation"].fillna(-1)
df_leaked["final_bill_amount"] = df_leaked["final_bill_amount"].fillna(-1)

leaked_features = legitmat_features + ["days_since_cancellation","retention_offer_after_cancellation","final_bill_amount"]

X_leaked=pd.get_dummies(df_leaked[leaked_features],columns=["contract_type"],drop_first=True)
y=df["churn"]

X_train, X_test, y_train, y_test = train_test_split(
   X_leaked,y,test_size=.25,random_state=42,stratify=y
)

model_leaked = LogisticRegression(max_iter=1000).fit(X_train,y_train)

acc_leaked = accuracy_score(y_test,model_leaked.predict(X_test))
auc_leaked = roc_auc_score(y_test,model_leaked.predict_proba(X_test)[:,1])

print(acc_leaked*100)
auc_leaked*100
```
